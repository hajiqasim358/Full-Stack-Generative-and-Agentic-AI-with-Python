```python
# Import libraries and load the API key

import os
from pathlib import Path
from dotenv import load_dotenv
from groq import Groq

env_path = Path.cwd() / ".env"
load_dotenv(env_path, override=True)

api_key = os.getenv("GROQ_API_KEY")

if not api_key:
    raise ValueError("Groq API key not found.")

print("API key loaded successfully!")

```

```python
# Create the Groq Client

client = Groq(api_key=api_key)

print("Groq Clinet Initialized Successfully!")
```

```python
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "system",
            "content": "You are a friendly AI tutor teaching agentic AI to beginners."
        },
        {
            "role": "user",
            "content": "What is agentic AI? Explain with one simple example."
        }
    ],
    temperature=0.3
)

print("Response received successfully!")
```

```python
# Display the answer
answer = response.choices[0].message.content

print("AI Response:\n")
print(answer)
```
